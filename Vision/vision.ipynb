# Vision IA

```python
from ultralytics import YOLO
import cv2
```

```python
model = YOLO("yolo11n.pt")  # léger et rapide
print("Modèle YOLO chargé avec succès.")

# Ouvre la webcam
cap = cv2.VideoCapture(0)

if not cap.isOpened():
    print("Erreur : impossible d'ouvrir la webcam.")
    input("Appuyez sur cv pour quitter.")
 

# Classe COCO "person" = 0
PERSON_CLASS_ID = 0

while True:
    ret, frame = cap.read()
    if not ret:
        print("Erreur : impossible de lire l'image webcam.")
        input("Appuyez sur cv pour quitter.")

    # Inférence YOLO
    results = model.predict(
        source=frame,
        conf=0.5,
        verbose=False
    )

    annotated = frame.copy()

    for result in results:
        boxes = result.boxes

        if boxes is None:
            continue

        for box in boxes:
            cls = int(box.cls[0].item())
            conf = float(box.conf[0].item())

            if cls == PERSON_CLASS_ID:
                x1, y1, x2, y2 = map(int, box.xyxy[0].tolist())

                cv2.rectangle(annotated, (x1, y1), (x2, y2), (0, 255, 0), 2)
                label = f"person {conf:.2f}"
                cv2.putText(
                    annotated,
                    label,
                    (x1, max(y1 - 10, 0)),
                    cv2.FONT_HERSHEY_SIMPLEX,
                    0.7,
                    (0, 255, 0),
                    2
                )

    cv2.imshow("Detection de personne - YOLO", annotated)

    # Quitter avec q
    if cv2.waitKey(1) & 0xFF == ord("q"):
        break

cap.release()
cv2.destroyAllWindows()
```
